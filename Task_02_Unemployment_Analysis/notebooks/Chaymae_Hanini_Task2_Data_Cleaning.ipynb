```python
import pandas as pd
import numpy as np
```

```python
#Load data :
```

```python
df = pd.read_csv("../data/Unemployment_in_India.csv")
```

```python
#statistic description of data:
```

```python
print(df.head())
print(df.info())
print(df.describe())
print("Missing values per column:\n", df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())
 
```

```python
df[df.isnull().all(axis=1)]
```

```python
#handling missing value & duplicates row
```

```python
df = df.dropna(how='all')
```

```python
#verify:
print(df.shape)

print(df.isnull().sum())
```

```python
#check the type for each column:
print(df.info())
```

```python
#verify duplicates row:
print(df.duplicated().sum())
```

```python
#remove the space name of columns:
df.columns = df.columns.str.strip()
```

```python
#transforme date object to datetime :
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)
```

```python
print(df.info())
```

```python
#Feature Engineering:
```

```python
#year:
df["Year"] = df["Date"].dt.year
#Month:
df["Month"] = df["Date"].dt.month
```

```python
print(df.info())
```

```python
#save data to Unemployment_Cleaned.csv:
df.to_csv("../data/Unemployment_Cleaned.csv", index=False)
```
