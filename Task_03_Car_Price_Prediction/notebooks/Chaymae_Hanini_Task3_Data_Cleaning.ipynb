```python
import pandas as pd
import numpy as np
```

```python
#Load data :
```

```python
df = pd.read_csv("../data/cardekho.csv")
```

```python
#statistic description of data:
```

```python
print(df.head())
print(df.info())
print(df.describe())
print("Missing values per column:\n", df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
duplicate_percentage = df.duplicated().mean() * 100
print(f"Duplicate percentage: {duplicate_percentage:.2f}%")
```

```python
#convert column max_power type object to type float
```

```python
df['max_power'].unique()[:30]
```

```python
df['max_power'].nunique()
```

```python
df['max_power'] = (
    df['max_power']
    .str.replace('bhp', '', regex=False)
    .str.strip()
)

df['max_power'] = pd.to_numeric(df['max_power'], errors='coerce')
```

```python
#check if it's work:
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
duplicates = df[df.duplicated(keep=False)]

print("Number of duplicate rows:", len(duplicates))
duplicates
```

```python
#handling duplicates:
```

```python
print("Before:", df.shape)

df = df.drop_duplicates()

print("After:", df.shape)
```

```python
#handling missing value: 
```

```python
print(df.isnull().sum())
```

```python
numeric_cols = [
    'mileage(km/ltr/kg)',
    'engine',
    'max_power',
    'seats'
]

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
print(df.isnull().sum())
```

```python
categorical_cols = [
    'fuel',
    'seller_type',
    'transmission',
    'owner'
]

for col in categorical_cols:
    print(f"\n{col}:")
    print(df[col].value_counts())
```

```python
print("Mileage = 0:", (df['mileage(km/ltr/kg)'] == 0).sum())

print("Engine = 0:", (df['engine'] == 0).sum())

print("Max power = 0:", (df['max_power'] == 0).sum())

print("Seats <= 0:", (df['seats'] <= 0).sum())

print("Seats > 10:", (df['seats'] > 10).sum())

print("KM driven > 1,000,000:",
      (df['km_driven'] > 1_000_000).sum())
```
